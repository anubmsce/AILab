```python
# Initialize the Tic Tac Toe board as a 3x3 grid of empty spaces
board = [[' ' for _ in range(3)] for _ in range(3)]

# Place 'X' and 'O' markers on the board according to the initial state provided by the user.
# Note: User provided 1-indexed coordinates, which are converted to 0-indexed here.
# X at (1,3) -> (0,2), (2,1) -> (1,0), (3,1) -> (2,0)
# O at (1,1) -> (0,0), (3,2) -> (2,1), (3,3) -> (2,2)
board[0][2] = 'X'
board[1][0] = 'X'
board[2][0] = 'X'

board[0][0] = 'O'
board[2][1] = 'O'
board[2][2] = 'O'

# Function to print the current state of the Tic Tac Toe board
def print_board(board):
    for row in board:
        print('|'.join(row))
        print('-' * 5)

# Function to check if a player has won the game
def check_win(board, player):
    # Check rows for a win
    for row in board:
        if all([cell == player for cell in row]):
            return True
    # Check columns for a win
    for col in range(3):
        if all([board[row][col] == player for row in range(3)]):
            return True
    # Check diagonals for a win
    if (board[0][0] == player and board[1][1] == player and board[2][2] == player) or \
       (board[0][2] == player and board[1][1] == player and board[2][0] == player):
        return True
    return False

# Function to check for a draw (no empty spaces left)
def check_draw(board):
    for row in board:
        for cell in row:
            if cell == ' ':
                return False
    return True

# Function to calculate the 'path cost' for the current board state.
# In this context, path cost is defined as the total number of moves (X's and O's) made.
def calculate_path_cost(board):
    moves = 0
    for row in board:
        for cell in row:
            if cell != ' ':
                moves += 1
    return moves

print("Welcome to Tic Tac Toe!")

current_player = 'X' # Based on the initial board state (3 X's, 3 O's), it's X's turn

# Game loop
while True:
    print("\nCurrent Board:")
    print_board(board)

    current_path_cost = calculate_path_cost(board)
    print(f"Current Path Cost (total moves): {current_path_cost}")

    # Check if a player has already won or if it's a draw from the previous turn (useful for the initial state check)
    # The previous check was removed, so this will only check after a player's move.

    move_made = False
    while not move_made:
        try:
            print(f"Player {current_player}'s turn.")
            row = int(input("Enter row (1-3): ")) - 1
            col = int(input("Enter column (1-3): ")) - 1

            if 0 <= row < 3 and 0 <= col < 3:
                if board[row][col] == ' ':
                    board[row][col] = current_player
                    move_made = True
                else:
                    print("That cell is already taken! Try again.")
            else:
                print("Invalid row or column. Please enter numbers between 1 and 3.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    # Check for win after the current player's move
    if check_win(board, current_player):
        print("\nFinal Board:")
        print_board(board)
        final_path_cost = calculate_path_cost(board)
        print(f"Current Path Cost (total moves): {final_path_cost}")
        print(f"Player {current_player} wins! Congratulations!")
        break

    # Check for draw after the current player's move if no one has won
    if check_draw(board):
        print("\nFinal Board:")
        print_board(board)
        final_path_cost = calculate_path_cost(board)
        print(f"Current Path Cost (total moves): {final_path_cost}")
        print("It's a draw!")
        break

    # Switch player for the next turn
    current_player = 'O' if current_player == 'X' else 'X'
```
