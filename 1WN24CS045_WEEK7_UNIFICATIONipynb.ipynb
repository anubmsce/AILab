```python

def occurs_check(var, expr, subst):
    expr = apply_substitution(expr, subst)

    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs_check(var, arg, subst) for arg in expr[1])

    return False


def apply_substitution(expr, subst):
    if isinstance(expr, str):
        while expr in subst:
            expr = subst[expr]
        return expr

    if isinstance(expr, tuple):
        return (expr[0], tuple(apply_substitution(arg, subst) for arg in expr[1]))

    return expr


def unify(x, y, subst):
    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    if x == y:
        return subst

    if isinstance(x, str) and x[0].isupper():
        if occurs_check(x, y, subst):
            return None
        subst[x] = y
        return subst

    if isinstance(y, str) and y[0].isupper():
        if occurs_check(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x[1]) != len(y[1]):
            return None

        for a, b in zip(x[1], y[1]):
            subst = unify(a, b, subst)
            if subst is None:
                return None

        return subst

    return None


def parse(expr):
    import ast

    node = ast.parse(expr, mode="eval").body

    def convert(n):
        if isinstance(n, ast.Name):
            return n.id

        if isinstance(n, ast.Call):
            return (n.func.id, tuple(convert(arg) for arg in n.args))

        raise ValueError("Invalid expression")

    return convert(node)


def main():
    print("UNIFICATION ALGORITHM")

    expr1 = input("Enter first expression: ")
    expr2 = input("Enter second expression: ")

    try:
        x = parse(expr1)
        y = parse(expr2)

        result = unify(x, y, {})

        if result is None:
            print("Unification Failed")
        elif not result:
            print("Expressions are already identical.")
            print("Substitution: NIL")
        else:
            print("Unification Successful")
            print("Substitution Set:")
            for key, value in result.items():
                print(key, "=", value)

    except Exception:
        print("Invalid Expression")


if __name__ == "__main__":
    main()
```
