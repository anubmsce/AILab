```python

# PROOF BY RESOLUTION USING FIRST ORDER PREDICATE LOGIC

# Literal format:
# (sign, predicate, arguments)
# True = Positive
# False = Negative
print("1WN24CS045 ANU N")
def is_variable(x):
    return x.islower()

def unify(x, y, subs):
    if x == y:
        return subs

    if is_variable(x):
        if x in subs:
            return unify(subs[x], y, subs)
        subs[x] = y
        return subs

    if is_variable(y):
        if y in subs:
            return unify(x, subs[y], subs)
        subs[y] = x
        return subs

    return None


def apply_substitution(literal, subs):
    sign, predicate, args = literal
    new_args = []

    for arg in args:
        while arg in subs:
            arg = subs[arg]
        new_args.append(arg)

    return (sign, predicate, tuple(new_args))


def resolve(c1, c2):
    for i, l1 in enumerate(c1):
        for j, l2 in enumerate(c2):

            if l1[1] == l2[1] and l1[0] != l2[0]:

                subs = {}

                for a, b in zip(l1[2], l2[2]):
                    subs = unify(a, b, subs)

                    if subs is None:
                        break

                if subs is not None:
                    result = []

                    for k, lit in enumerate(c1):
                        if k != i:
                            result.append(
                                apply_substitution(lit, subs)
                            )

                    for k, lit in enumerate(c2):
                        if k != j:
                            result.append(
                                apply_substitution(lit, subs)
                            )

                    return result, subs

    return None, None


def display_clause(clause):
    if not clause:
        return "{}"

    result = []

    for sign, predicate, args in clause:
        text = predicate + "(" + ", ".join(args) + ")"

        if not sign:
            text = "¬" + text

        result.append(text)

    return " ∨ ".join(result)


# ==========================================
# STEP 1: KNOWLEDGE BASE
# ==========================================

C1 = [(False, "food", ("x",)),
      (True, "likes", ("John", "x"))]

C2 = [(True, "food", ("Apple",))]

C3 = [(True, "food", ("vegetables",))]

C4 = [(False, "eats", ("y", "z")),
      (True, "killed", ("y",)),
      (True, "food", ("z",))]

C5 = [(True, "eats", ("Anil", "Peanuts"))]

C6 = [(True, "alive", ("Anil",))]

C7 = [(False, "eats", ("Anil", "w")),
      (True, "eats", ("Harry", "w"))]

C8 = [(True, "killed", ("g",)),
      (True, "alive", ("g",))]

C9 = [(False, "alive", ("k",)),
      (False, "killed", ("k",))]

C10 = [(True, "likes", ("John", "Peanuts"))]

# Negated goal
C11 = [(False, "likes", ("John", "Peanuts"))]


# ==========================================
# STEP 2: DISPLAY KNOWLEDGE BASE
# ==========================================

print("KNOWLEDGE BASE")
print("-------------------------------")

clauses = [C1, C2, C3, C4, C5, C6, C7, C8, C9, C10]

for i, clause in enumerate(clauses, 1):
    print(f"C{i}: {display_clause(clause)}")

print("\nNegated Goal:")
print("C11:", display_clause(C11))


# ==========================================
# STEP 3: ACTUAL RESOLUTION
# ==========================================

print("\n\nPROOF BY RESOLUTION")
print("===============================")

# Step 1: Resolve C4 and C5

R1, S1 = resolve(C4, C5)

print("\nStep 1: Resolve C4 and C5")
print("Substitution:", S1)
print("Result:", display_clause(R1))


# Step 2: Resolve C9 and C6

R2, S2 = resolve(C9, C6)

print("\nStep 2: Resolve C9 and C6")
print("Substitution:", S2)
print("Result:", display_clause(R2))


# Step 3: Resolve R1 and R2

R3, S3 = resolve(R1, R2)

print("\nStep 3: Resolve Step 1 and Step 2")
print("Substitution:", S3)
print("Result:", display_clause(R3))


# Step 4: Resolve C1 and R3

R4, S4 = resolve(C1, R3)

print("\nStep 4: Resolve C1 and Step 3")
print("Substitution:", S4)
print("Result:", display_clause(R4))


# Step 5: Resolve R4 and C11

R5, S5 = resolve(R4, C11)

print("\nStep 5: Resolve Step 4 and Negated Goal")
print("Substitution:", S5)
print("Result:", display_clause(R5))


# ==========================================
# STEP 4: FINAL RESULT
# ==========================================

print("\n===============================")

if R5 == []:
    print("EMPTY CLAUSE: {}")
    print("Hence proved!")
    print("John likes Peanuts.")
else:
    print("Proof could not be completed.")


# ==========================================
# STEP 5: RESOLUTION TREE DIAGRAM
# ==========================================

print("\n\n          RESOLUTION TREE")
print("          --------------\n")

print("     C4                    C5")
print("      \\                  /")
print("       \\                /")
print("        R1: " + display_clause(R1))

print("\n     C9                    C6")
print("      \\                  /")
print("       \\                /")
print("        R2: " + display_clause(R2))

print("\n        R1                R2")
print("         \\              /")
print("          \\            /")
print("           R3: " + display_clause(R3))

print("\n     C1                    R3")
print("      \\                  /")
print("       \\                /")
print("        R4: " + display_clause(R4))

print("\n     R4                    C11")
print("      \\                  /")
print("       \\                /")
print("          R5: " + display_clause(R5))

print("\n          HENCE PROVED!")
print("      John likes Peanuts.")
```
